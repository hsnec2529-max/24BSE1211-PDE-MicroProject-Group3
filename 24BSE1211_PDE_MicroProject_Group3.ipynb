###**Objective**
Simulate the standing wave behavior and probability density functions $|\Psi_n(x,t)|^2$ for energy states $n = 1, 2, 3$
.

```python
import matplotlib.pyplot as plt
import numpy as np

# Domain and constants
L = 1.0  # Well length (e.g., 1 nm normalized)
x = np.linspace(0, L, 1000)


def psi_n(n, x_val, L_val=1.0):
  return np.sqrt(2.0 / L_val) * np.sin(n * np.pi * x_val / L_val)


def prob_density(n, x_val, L_val=1.0):
  return (psi_n(n, x_val, L_val)) ** 2


# Plotting wavefunctions and densities
fig, axes = plt.subplots(3, 2, figsize=(12, 9), sharex=True)
colors = ['#1f77b4', '#ff7f0e', '#2ca02c']
n_levels = [1, 2, 3]

for i, n in enumerate(n_levels):
  # Wavefunction psi_n(x)
  axes[i, 0].plot(
      x,
      psi_n(n, x, L),
      color=colors[i],
      lw=2,
      label=rf'$\%psi_{n}(x) (n={n})$',
  )
  axes[i, 0].axhline(0, color='black', linestyle='--', alpha=0.4)
  axes[i, 0].set_ylabel('Amplitude')
  axes[i, 0].grid(True, linestyle=':', alpha=0.6)
  axes[i, 0].legend(loc='upper right')

  # Probability density |psi_n(x)|^2
  prob = prob_density(n, x, L)
  axes[i, 1].plot(
      x, prob, color=colors[i], lw=2, label=rf'$|\Psi_{n}(x)|^2 (n={n})$'
  )
  axes[i, 1].fill_between(x, prob, color=colors[i], alpha=0.25)
  axes[i, 1].set_ylabel('Probability Density')
  axes[i, 1].grid(True, linestyle=':', alpha=0.6)
  axes[i, 1].legend(loc='upper right')

axes[0, 0].set_title(
    r'Stationary Wavefunctions $\psi_n(x)$',
    fontweight='bold'
)
axes[0, 1].set_title(
    r'Probability Densities $|\Psi_n(x)|^2$',
    fontweight='bold'
)
axes[2, 0].set_xlabel('Position $x$ / $L$')
axes[2, 1].set_xlabel('Position $x$ / $L$')

plt.tight_layout()
plt.show()
```
